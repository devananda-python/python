```python
#prgrm to get the 3rd element and 3rd element from last of a tuple.
t=tuple(map(int,input("enter the number:").split(",")))
print("3rd element:",t[2])
print("3rd  element from last:",t[-3])
```

```python
#prgrm to accepts a sequence of comma-separated numbers from user and generate a tuple with those numbers and find average of all items in that tuple
numbers=tuple(map(int,input("enter comma_separated numbers:").split(",")))
print("tuple:",numbers)
print("average:",sum(numbers)/len(numbers))





```
