```python
#sort dictionary ascending or descending order
roll=input("enter roll numbers:").split(" ")
name=input("enter name of students:").split(" ")
my_dict=dict(zip(roll,name))
print("Original dictionary is:",my_dict)
print("ascending order:",dict(sorted(my_dict.items())))
print("descending order:",dict(sorted(my_dict.items(),reverse=True)))
```

```python
#merge 2 dictionaries and display it
dict1=dict(zip(input("enter 3 roll numbers:").split(),input("enter 3 names:").split()))
dict2=dict(zip(input("enter 2 roll numbers:").split(),input("enter 2 names:").split()))
print("dictionary1:",dict1)
print("dictionary2:",dict2)
dict1.update(dict2)
print("merged dictionary:",dict1)      
```
