```python
#generate a positive list of numbers
a=[-2,5,-3,3,2,-6,-1,1]
list=[i for i in a if i>0]
print(list)
```

```python
#find squares of N numbers
n=int(input("enter N:"))
square=[i**2 for i in range(1,n+1)]
print(square)
```

```python
#form a list of vowels selected from a given word
word=(input("enter a word:"))
vowels=[i for i in word if i in "aeiou"]
print(vowels)
```

```python
#list the ordinal value of each element using(ord())
s=input("enter a string:")
values=[ord(i) for i in s]
print(values)
```
