```python
#program to printout all colors from colorset1 not contained in colorset2.
set1=set(input("enter colors in set1:").split(","))
set2=set(input("enter colors in set2:").split(","))
set3=set1-set2
print(set3)
```

```python
#program to printout all colors from color set1 and color set 2.
set1=set(input("enter colors in set1:").split(","))
set2=set(input("enter colors in set2:").split(","))
set3=set1|set2
print(set3)
```
